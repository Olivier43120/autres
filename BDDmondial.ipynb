# Base de données Mondial — exercices SQL

La base `mondial.db` est chargée par Basthon grâce au paramètre `aux` de l'URL.
La première cellule SQL l'attache sous le nom `mondial`.

**Important :** ce notebook doit être ouvert avec `kernel=sql` dans l'URL Basthon. Il ne faut pas double-cliquer sur le fichier `.ipynb` pour le tester.

````sql
ATTACH DATABASE 'mondial.db' AS mondial;
````

## Vérification du chargement

Cette requête liste les tables présentes dans `mondial.db`.

````sql
SELECT name
FROM mondial.sqlite_master
WHERE type = 'table'
ORDER BY name;
````

## Exemple de requête

Les tables de la base sont accessibles avec le préfixe `mondial.`.
Par exemple :

```sql
SELECT *
FROM mondial.Country
LIMIT 5;
```


````sql
SELECT *
FROM mondial.Country
LIMIT 5;
````
