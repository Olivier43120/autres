```python
import sqlite3
import pandas as pd
from pyodide.http import pyfetch

# Téléchargement de la base de données SQLite
url = "https://raw.githubusercontent.com/Olivier43120/autres/main/mondial.db"

response = await pyfetch(url)

if response.status == 200:
    content = await response.bytes()
    with open("mondial.db", "wb") as f:
        f.write(content)
    print("Base de données chargée avec succès !")
    
    # Connexion à la base
    conn = sqlite3.connect("mondial.db")

    # Requête SQL de test
    df = pd.read_sql_query("SELECT * FROM Country LIMIT 10;", conn)
    display(df)
else:
    print(f"Erreur lors du téléchargement. Code HTTP : {response.status}")
```
